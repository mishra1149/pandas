```python
import numpy as np
import pandas as pd
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
df
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

```python
df
```

```python
df.dropna() #works on the rows wise check each row and delete the row if there is an null value
```

```python
df.dropna(thresh=4) #itna not null value hona chy  usko rakhke delete kar dega 
```

```python
df.dropna(thresh=3) #3 not null value ho usko rkhke delete kr dega sav 
```

```python
values={'A':0,'B':100,'C':200,'D':900}
df.fillna(value=values,inplace=True) #yaha par permanentlty data save ho gaya 
```
